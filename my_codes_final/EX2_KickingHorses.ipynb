### 2. Kicking horses

A famous early  application of low-number statistics was an analysis of Prussian cavalryman horse-kick deaths by [Bortkiewicz](https://www.wikiwand.com/en/Ladislaus_Bortkiewicz) in 1898.

He studied the distribution of 122 men kicked to death by horses among 10 Prussian army corps within 20 years (so a total number of corpes of 200). He recorded the number of corps with a given number of deaths:

| Number of deaths | Number of groups |
| --- | --- |
| 0 | 109 |
| 1 | 65 |
| 2 | 22 |
| 3 | 3 |
| 4 | 1 |

- Plot the resulting probability distribution. Careful with the normalization
- How does it look like? Again try to fit it by hand.

(Hint: think about sample mean and sample variance)


Ha seguito 10 gruppi, ciascun gruppo per ogni anno ha dato il suo registro, ovvero quanti ne sono morti per horse-kicking, questo per 20 anni, dunque ciascun gruppo ha dato 20 "registri" totalmente. Presi questi 200 registri si è visto in quanti di questi registri annuali ci fossero 1 morte, 2 morti, 0 morti etc... Difatti sommando la colonna a destra ho 200 registri e invece moltiplicando i registri per il numero di morti ottengo esattamente 122

```python
import numpy as np
import scipy.stats
import pylab as plt
```

```python
deaths = [0, 1, 2, 3, 4] #il numero di morti nel registro
groups = [109, 65, 22, 3, 1] #quanti registri hanno riportato il numero di morti corrispondente
deaths = np.array(deaths)
groups = np.array(groups)
total_groups = np.sum(groups)
#frequenza relativa della morte nei gruppi
freq = groups/total_groups
freq
#Freq è un array che contiene la frequenza relativa di ciascun numero di morti nel totale dell'osservazione
plt.scatter(deaths,freq)
```

```python
#Facciamo la media campionaria dell'evento "uccisione per horse-kicking". In questo caso facciamo una media pesata perché sono già organizzati i dati per "quante volte si è manifestato", se fosse stata una lista bastava np.mean
sample_mean = np.average(deaths, weights = freq)
print(sample_mean) #Questa è la media della Poissoniana, ogni quanto si manifesta una morte per horse-kicking?
```

```python
#Facciamo un fit
p = scipy.stats.poisson(sample_mean)
plt.scatter(deaths, p.pmf(deaths), label = 'Distribuzione fittata sui dati') #nota, la pmf è l'equivalente della pdf per distribuzioni discrete
plt.bar(deaths,freq, label = 'Dati', alpha = 0.3)
plt.legend()
print(deaths*groups)
```
